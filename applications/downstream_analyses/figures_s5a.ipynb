## Moffitt2018

```python
scrna_file = '/localscratch/mlobo6/spadecoder/datasets/dataset1_merfish_moffitt2018/data/scrna_ref_normnolog_nov2024_1k.h5ad'

adata = sc.read(scrna_file)

scrna_celltype = 'Cell class (determined from clustering of all cells)'
```

```python
sc.pl.umap(adata, add_outline=True,frameon=False,color=scrna_celltype,title='Moffitt2018 scRNAseq')
```

```python
sc.pl.umap(adata, add_outline=True,frameon=False,color=scrna_celltype,title='Moffitt2018 scRNAseq',legend_loc=None)
```
