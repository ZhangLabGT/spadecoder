```python

spa_celltype_key='cell_type'

dataset = 'Moffitt2018'

scrna_cluster_key = "Cell class (determined from clustering of all cells)"

deconv_dir = figdir + dataset + '/deconv/'
metrics_dir = figdir + dataset + '/metrics/'


N = 50
nneigh = 10
nbdswaps = 2
par_lambda_curr = 0.1
# par_eta1_curr = 10.0
# kernel3d_bw_slices_curr = 8
bandwidth_curr = 0.01
n_spatial_neigh_curr = 10
mode_nbd = 'variabletranscr'
gtalign = False # if True, use GT alignment 
aligntool = 'moscot'
key_name = 'linear_0'
# num_augment = '20'
mode_nbd = 'variabletranscr'
augment = True
mode = 'multislice'
sc_type = 'sc'

all_algos = {}

for num_augment in [5, 10, 15, 20]:
    for kernel3d_bw_slices_curr in [2, 4, 8, 16]: 
        partesting_str =     '_partesting_' + 'parlambda_' + str(par_lambda_curr) +   '_parbw_' + str(bandwidth_curr) +  '_nbdtype_' + mode_nbd + '_gtalign_' + str(gtalign) + '_augment_' + str(augment) + '_kernel3d_bw_slices_' + str(kernel3d_bw_slices_curr) + '_num_augment_' + str(num_augment) + '_realalign_' + str(aligntool) 
        res_file = 'metrics_' + mode + '_' + mode_nbd + '_sptsz_' + str(N)  + '_nneigh_' + str(nneigh) + '_nbdswaps_' + str(nbdswaps) + partesting_str +'_' + sc_type + '_sim.pickle'
        print(res_file)
        with open(metrics_dir + res_file, 'rb') as handle:
            metrics_sc_ms = pickle.load(handle)
        all_algos[ 'Aug-' + str(num_augment) + '-3DBW-' + str(kernel3d_bw_slices_curr)] = metrics_sc_ms




result_metrics = ['orig_rmse_avg',  'avg_corr_pe_avg', 'avg_jsd_avg']
col_names = result_metrics.copy()
col_names.extend(['algo'])
res_df = pd.DataFrame(columns=col_names)
idx = 0



for algo_name, algo_res in all_algos.items():
    real_samp = list(algo_res.keys())
    # sim_samp = metrics_sc_ms[entry][real_samp[0]].keys()
    for real_samp_name in real_samp:

        tmp_df = algo_res[real_samp_name].loc[result_metrics,:].T
        tmp_df['algo'] = algo_name
        tmp_df['sample'] = real_samp_name
        res_df = pd.concat([res_df,tmp_df])


        
res_df['dataset'] = 'Moffitt2018'
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def plot_sorted_rmse(df, algo_col, rmse_col,
                     save_path=None, xlabel="Algorithm", ylabel="RMSE", title="Sorted RMSE with Error Bars"):
    """
    Plots RMSE values for multiple algorithm versions, sorted by median RMSE.
    
    Parameters:
    - df: DataFrame with RMSE values
    - algo_col: column with algorithm names/versions
    - rmse_col: column with RMSE values
    """

    # Compute summary stats
    summary = df.groupby(algo_col)[rmse_col].agg(['mean', 'median', 'min', 'max']).reset_index()

    # Sort by median RMSE ascending
    if 'corr' in rmse_col: 
        summary = summary.sort_values(by='mean', ascending=False)
    else:
        summary = summary.sort_values(by='mean', ascending=True)
    algo_sorted = summary[algo_col].values

    # Reorder df to match sort
    df[algo_col] = pd.Categorical(df[algo_col], categories=algo_sorted, ordered=True)
    df = df.sort_values(by=algo_col)

    # X positions
    x = np.arange(len(algo_sorted))

    # Plot
    fig, ax = plt.subplots(figsize=(max(16, len(algo_sorted)*0.7), 8))

    # Plot individual points
    for i, algo in enumerate(algo_sorted):
        vals = df[df[algo_col] == algo][rmse_col]
        ax.scatter([i] * len(vals), vals, alpha=0.6, s=30, color="gray", edgecolor="k", linewidth=0.3)

    # Plot error bars (min-max) and mean
    means = summary['mean'].values
    # lower = means - summary['min'].values
    # upper = summary['max'].values - means
    # ax.errorbar(x, means, yerr=[lower, upper], fmt='o', capsize=6, color='crimson', label='Mean ± range')
    # Plot mean line
    ax.plot(x, means, marker='o', linestyle='-', color='red', linewidth=2, label='Mean ' + result_metric_newname[metric] )

    # Annotate mean values
    y_max = ax.get_ylim()[1]
    for xi, yi in zip(x, means):
        offset = 0.01 * (y_max)  # dynamic vertical offset
        text_y = yi + offset
        if text_y > y_max * 0.98:
            text_y = yi  # keep it right on the point if too high
        ax.text(xi, text_y, f'{yi:.3f}', ha='center', va='bottom', fontsize=24)

    # Format
    ax.set_xticks(x)
    ax.set_xticklabels(algo_sorted, rotation=45, ha='right', fontsize=24)
    ax.tick_params(axis='y', labelsize=24) 
    ax.set_ylabel(ylabel, fontsize=32)
    # ax.set_xlabel(xlabel, fontsize=14)
    ax.set_title(title, fontsize=32)
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.legend()

    ax.legend(fontsize=24, loc='best', frameon=True, shadow=True, 
              fancybox=True, framealpha=0.9, edgecolor='black', 
              markerscale=1.5)
    
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=1200, bbox_inches='tight')
    plt.show()


dataset = [  'Moffitt2018', # ,'Moffitt2018-high',
           'Choi2023',# ,'Choi2023-high',# 'dataset3_seqfishpluscai', 'dataset4_seqfishpluscai',  'dataset5_starmap_mousevisualcortex',
            'Haviv2025']
result_metric_newname = {'orig_rmse_avg':'RMSE','avg_corr_pe_avg':'Aver Corr','avg_jsd_avg':'Aver JSD'}


for dataset_curr in dataset:
    for metric in result_metrics:
        res_df_tmp = res_df[res_df['dataset']==dataset_curr]
        tmp_plot_name = result_metric_newname[metric]
        tmp_plot_name = tmp_plot_name.replace(' ','_')
        plot_sorted_rmse(
            df=res_df_tmp,                    # your DataFrame
            algo_col='algo', # e.g., 'algo'
            rmse_col=metric,              # e.g., 'orig_rmse_avg'
            
            save_path=figdir + 'sorted_' + dataset_curr + ':' + tmp_plot_name + '.pdf',
            ylabel=result_metric_newname[metric],
           title= dataset_curr + ':' + result_metric_newname[metric] 
        )
```
