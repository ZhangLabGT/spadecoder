```python
def add_to_adata(tensor, adata, celltypes, prefix):
    arr = np.array(tensor)
    arr_2d = arr.reshape(arr.shape[0], -1)
    col_labels = [prefix + entry + '_' + entry2 for entry in celltypes for entry2 in celltypes]
    arr_2d = pd.DataFrame(arr_2d, columns=col_labels)

    for entry in col_labels:
        adata.obs[entry] = list(arr_2d[entry].values)
    
    return adata
```

```python
dataset_full = 'dataset11_thymus_curated_TECsubsets'
suffix = '_norm1knolog'
adata_spa_path = '../../datasets/' + dataset_full + '/data/spatial_norm.pickle'

# ground truth 
with open(adata_spa_path, 'rb') as handle:
    adata_spa = pickle.load(handle)
```

```python
import anndata 
adata_tmp_dict = {} 
for entry in ['16thpcw', '17thpcw', '18thpcw', '19thpcw']:
    adata_tmp_dict[entry] = adata_spa[entry].copy()
    adata_tmp_dict[entry].obs['batch'] = str(entry)


adata_concat = anndata.concat(adata_tmp_dict)

adata_concat.obs['batch'] = adata_concat.obs['batch'].astype('category')

import moscot as mt
from moscot import datasets
from moscot.problems.space import MappingProblem
from moscot.problems.space import AlignmentProblem
from moscot.problems.generic import FGWProblem


ap = AlignmentProblem(adata=adata_concat)
ap = ap.prepare(batch_key="batch", policy="sequential")
ap = ap.solve()
ap.align(reference='16thpcw', mode='affine', key_added="seq_warp")
```

```python
adata_concat = add_to_adata(local_sci, adata_concat, celltypes, 'local_sci_')
adata_concat = add_to_adata(local_sci_pvals, adata_concat, celltypes, 'local_sci_pvals_')
adata_concat = add_to_adata(local_sci_pvals_both, adata_concat, celltypes, 'local_sci_pvals_both_')
adata_concat = add_to_adata(local_sci_sig_celltypes, adata_concat, celltypes, 'local_sci_sig_celltypes_')
adata_concat = add_to_adata(local_sci_sig_celltypes_both, adata_concat, celltypes, 'local_sci_sig_celltypes_both_')
```
