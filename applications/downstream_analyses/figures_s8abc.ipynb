```python
import pickle
import os 
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import numpy as np
from scipy.stats import pearsonr, spearmanr
from scipy.spatial.distance import cdist
import seaborn as sns
```

```python
plt.rcParams['figure.figsize']=(8,8) #rescale figures
sc.settings.verbosity = 3
sc.logging.print_versions()
import os
import sys
sc.set_figure_params(scanpy=True, dpi_save=400,dpi=150)
import matplotlib
```

```python
figdir = '../fig2_supp9b_varying_spot_sz/'
```

```python

dataset = 'Moffitt2018'
dataset_full = 'dataset1_merfish_moffitt2018'
pickle_path = '../../datasets/' + dataset_full + '/results/simulations/pickles/'
N_curr = 50
nneigh = 10
nswaps_nbd = 2
par_bw = 0.01
suffix = '_normnolog_nov2024_1k'
aligntool = 'moscot'
mode = 'multislice'
gtalign = False
transcrnbd = 'variabletranscr'

num_spots = {}
for N_curr in [5, 10, 25, 50, 75, 100]:
    adata_spa_path = pickle_path + 'multi_slice_simulated_' + 'sptsz_' + str(N_curr)  + '_nneigh_' + str(nneigh) + '_nbdswaps_' + str(nswaps_nbd) + suffix + '.pickle' # '_old.pickle'        
    with open(adata_spa_path, 'rb') as handle:
        adata_spa = pickle.load(handle)
    num_spots[N_curr] = []
    for realslice in adata_spa['linear_0'][0].keys():
        for simslice in adata_spa['linear_0'].keys():
            num_spots[N_curr].extend(adata_spa['linear_0'][simslice][realslice].obs.sum(axis=1).values.tolist())
```

```python
figdir
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set a consistent style
sns.set_theme(style="whitegrid", context="talk")

save_dir = "../fig2_supp9b_varying_spot_sz/"

sns.set_theme(style="whitegrid", context="talk")

for spt_sz, values in num_spots.items():
    plt.figure(figsize=(8, 5))
    
    # Main histogram
    sns.histplot(
        values, bins=40, color="#4C72B0", edgecolor="white",
        alpha=0.85, kde=True, linewidth=1.2
    )

    # Add median line for context
    plt.axvline(
        np.median(values), color="#E15759", linestyle="--",
        linewidth=2, label=f"Median = {np.median(values):.1f}"
    )
    
    # Titles and labels
    plt.title(f"{dataset} | N = {spt_sz}", fontsize=30, weight='bold', pad=15)
    plt.xlabel("Number of Cells per Spot", fontsize=30)
    plt.ylabel("Frequency", fontsize=30)

    # Tick aesthetics
    plt.xticks(fontsize=24)
    plt.yticks(fontsize=30)
    
    # # Subtle frame cleanup
    # sns.despine(left=True, bottom=True)
    # plt.legend(frameon=False, fontsize=14)
    
    # plt.tight_layout()
    # plt.show()
    plt.tight_layout()

    # Save figure
    save_path = os.path.join(save_dir, f"{dataset}_N{spt_sz}_histogram.pdf")
    plt.savefig(save_path, dpi=1200, bbox_inches="tight", transparent=True)

    plt.show()
    print(f"Saved: {save_path}")
```
