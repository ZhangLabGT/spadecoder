```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Example data: random spatial coordinates
# np.random.seed(0)
# n_points = 1000
# x = np.random.rand(n_points)
# y = np.random.rand(n_points)
composition_df = {} 

ctypes = ['Astrocyte', 'Endothelial',
       'Ependymal', 'Excitatory', 'Immature oligodendrocyte', 'Inhibitory',
       'Mature oligodendrocyte', 'Microglia']

permutations  = {} 
df_single_index = {} 

nperm = 30

cnt = 0
for entry in slice_list: 
    points = entry.obsm['spatial'].copy()
    x = points[:,0]
    y = points[:,1]
    # Combine into (x, y)
    coords = points # np.stack([x, y], axis=1)

    # Define bin resolution
    n_bins_x = 6
    n_bins_y = 6

    # Bin edges
    x_edges = np.linspace(0, 1, n_bins_x + 1)
    y_edges = np.linspace(0, 1, n_bins_y + 1)

    # Compute bin indices for each point
    x_bin_idx = np.digitize(x, x_edges) - 1
    y_bin_idx = np.digitize(y, y_edges) - 1

    # # Optional: exclude out-of-bounds
    valid = (x_bin_idx >= 0) & (x_bin_idx <= n_bins_x) & (y_bin_idx >= 0) & (y_bin_idx <= n_bins_y)
    x_bin_idx = x_bin_idx[valid]
    y_bin_idx = y_bin_idx[valid]

    # Count number of points per bin
    bin_counts = np.zeros((n_bins_x+1, n_bins_y+1), dtype=int)
    for xb, yb in zip(x_bin_idx, y_bin_idx):
        bin_counts[xb, yb] += 1

    entry.obs['x_bin_idx'] = x_bin_idx
    entry.obs['y_bin_idx'] = y_bin_idx


    composition_df[str(cnt)] = entry.obs.groupby(['x_bin_idx','y_bin_idx']).mean()

    composition_df[str(cnt)] = composition_df[str(cnt)].fillna(0)

    df_single_index[str(cnt)] = composition_df[str(cnt)].reset_index()

    df_single_index[str(cnt)]['spot'] = df_single_index[str(cnt)]['x_bin_idx'].astype(str) + ' ' + df_single_index[str(cnt)]['y_bin_idx'].astype(str)
    

    del df_single_index[str(cnt)]['x_bin_idx']
    del df_single_index[str(cnt)]['y_bin_idx']

    cnt = cnt + 1

    
    
```

```python
## generate permutations 


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Example data: random spatial coordinates
# np.random.seed(0)
# n_points = 1000
# x = np.random.rand(n_points)
# y = np.random.rand(n_points)
composition_df = {} 

ctypes = ['Astrocyte', 'Endothelial',
       'Ependymal', 'Excitatory', 'Immature oligodendrocyte', 'Inhibitory',
       'Mature oligodendrocyte', 'Microglia']

permutations_dict  = {} 
# df_single_index = {} 

nperm = 100

cnt = 0
for entry_idx in range(len(slice_list)): 
    permutations_dict[entry_idx] = {} 

    adata_curr = slice_list[entry_idx].copy()
    # permute the cell-type p
    index_locs = list(adata_curr.obs.index)
    permutations = np.array([np.random.permutation(index_locs) for _ in range(nperm)])  # each row is a permutation of the indeices 
    obs_df = adata_curr.obs.copy()

    for perm_num in range(nperm):
        obs_df = obs_df.loc[permutations[perm_num,:],:]
    
        adata_curr.obs = obs_df.copy()

        # permutations_dict[entry_idx][str(perm_num)] = adata_curr

    


        points = adata_curr.obsm['spatial'].copy()
        x = points[:,0]
        y = points[:,1]
        # Combine into (x, y)
        coords = points # np.stack([x, y], axis=1)

        # Define bin resolution
        n_bins_x = 6
        n_bins_y = 6

        # Bin edges
        x_edges = np.linspace(0, 1, n_bins_x + 1)
        y_edges = np.linspace(0, 1, n_bins_y + 1)

        # Compute bin indices for each point
        x_bin_idx = np.digitize(x, x_edges) - 1
        y_bin_idx = np.digitize(y, y_edges) - 1

        # # Optional: exclude out-of-bounds
        valid = (x_bin_idx >= 0) & (x_bin_idx <= n_bins_x) & (y_bin_idx >= 0) & (y_bin_idx <= n_bins_y)
        x_bin_idx = x_bin_idx[valid]
        y_bin_idx = y_bin_idx[valid]

        # Count number of points per bin
        bin_counts = np.zeros((n_bins_x+1, n_bins_y+1), dtype=int)
        for xb, yb in zip(x_bin_idx, y_bin_idx):
            bin_counts[xb, yb] += 1

        adata_curr.obs['x_bin_idx'] = x_bin_idx
        adata_curr.obs['y_bin_idx'] = y_bin_idx


        permutations_dict[entry_idx][perm_num] = adata_curr.obs.groupby(['x_bin_idx','y_bin_idx']).mean()

        permutations_dict[entry_idx][perm_num]= permutations_dict[entry_idx][perm_num].fillna(0)

        permutations_dict[entry_idx][perm_num] = permutations_dict[entry_idx][perm_num].reset_index()

        permutations_dict[entry_idx][perm_num]['spot'] = permutations_dict[entry_idx][perm_num]['x_bin_idx'].astype(str) + ' ' + permutations_dict[entry_idx][perm_num]['y_bin_idx'].astype(str)
        

        del permutations_dict[entry_idx][perm_num]['x_bin_idx']
        del permutations_dict[entry_idx][perm_num]['y_bin_idx']

        

    # index_locs = list(df_single_index[str(entry)].index)

    # permutations = np.array([np.random.permutation(index_locs) for _ in range(nperm)])  # each row is a permutation of the indeices 

    

```

## JSD as distance - differential composition

```python
from scipy.stats import entropy

def jenson_shannon_distance(gnd_truth_norm, deconv_res):
    denom = (gnd_truth_norm + deconv_res)*0.5
    # cell_type_jsd = 0.5*entropy(pk=deconv_res_prop, qk=denom,axis=1)  +  0.5*entropy(pk=gnd_truth_norm_prop, qk=denom,axis=1) # .shape #  base=base)
    cell_type_jsd = 0.5*entropy(pk=deconv_res, qk=denom,axis=0)  +  0.5*entropy(pk=gnd_truth_norm, qk=denom,axis=0) # .shape #  base=base)
    # only compute average for cell types that are in both reference and query 
    # cell_type_jsd = cell_type_jsd[~np.isnan(cell_type_jsd)]
    # avg_jsd = cell_type_jsd.mean() 
    return np.sqrt(cell_type_jsd)
```

```python


jsd_dist = {}
jsd_perm_dist = {}

is_sig = {}
# diff_comp = []

nperm = 100


for sample_idx in range(11):
    for spt_num in range(df_single_index[str(sample_idx)].shape[0]):
        # compare same spots across adjacent slices 
        p1 = df_single_index[str(sample_idx)].loc[spt_num,ctypes].values
        p2 = df_single_index[str(sample_idx+1)].loc[spt_num,ctypes].values
                    
        jsd_perm_dist[str(sample_idx) + ' ' + str(spt_num)] = []
                    
        # p1 = ct_1.values[0]
        # p2 = ct_2.values[0]
        
        jsd_dist[str(sample_idx) + ' ' + str(spt_num)] = jenson_shannon_distance(p1.astype(float), p2.astype(float))
        

        for num_perm1 in range(nperm):
            for num_perm2 in range(nperm):
            
                p1 = permutations_dict[sample_idx][num_perm1].loc[spt_num,ctypes].values
                p2 = permutations_dict[sample_idx+1][num_perm2].loc[spt_num,ctypes].values
                
                
                jsd_tmp = jenson_shannon_distance(p1.astype(float), p2.astype(float))
                
                jsd_perm_dist[str(sample_idx) + ' ' + str(spt_num)].append(jsd_tmp)

                        
                        
```
