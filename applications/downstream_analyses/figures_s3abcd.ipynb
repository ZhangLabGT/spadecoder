```python

dataset = 'Moffitt2018'
dataset_full = 'dataset1_merfish_moffitt2018'
pickle_path = '../../datasets/' + dataset_full + '/results/simulations/pickles/'
N_curr = 50
nneigh = 10
nswaps_nbd = 2
par_bw = 0.01
suffix = '_normnolog_nov2024_1k'
aligntool = 'moscot'
mode = 'multislice'
gtalign = False
adata_spa_path = pickle_path + 'multi_slice_simulated_' + 'sptsz_' + str(N_curr)  + '_nneigh_' + str(nneigh) + '_nbdswaps_' + str(nswaps_nbd) + suffix + '.pickle' # '_old.pickle'        


# read spatial file 
if not os.path.exists(adata_spa_path):
    sys.exit() 
with open(adata_spa_path, 'rb') as handle:
    adata_spa = pickle.load(handle)



transcrnbd = 'variabletranscr'
kernel_path = '../fig2_supp5_fixed_vs_variable_neigh/' + dataset + '/kernel/kernel_wt_sptsz_' + str(N_curr) + '_nneigh_' + str(nneigh) + '_nbdswaps_' + str(nswaps_nbd) + '_parbw_' + str(par_bw) + suffix + '_' +  aligntool + '_' +  mode + '_' + transcrnbd + '_gtalign_' + str(gtalign) + '_sim.pickle'
with open(kernel_path, 'rb') as handle:
    kernels = pickle.load(handle)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


# Set a consistent style
sns.set_theme(style="whitegrid", context="talk")



kernel_all = []
for entry in kernels.keys():
    for entry2 in range(11):
        kernel_tmp = kernels[entry][str(entry2) + '_' + str(entry2)].copy()
        kernel_tmp = kernel_tmp[kernel_tmp > 0]
        kernel_all.extend(kernel_tmp.tolist())
        # np.fill_diagonal(kernel_tmp, 0.0)
        # kernel_tmp = (kernel_tmp > 0).sum(axis=0)
        # adata_spa['linear_0'][5][entry].obs['spaneigh'] = kernel_tmp
        # adata_spa['linear_0'][5][entry].obs['geary'] = gearys[entry]['5_5']['gearyc'].values
        # adata_spa['linear_0'][5][entry].obs['is_sig'] = gearys[entry]['5_5']['is_sig'].values
        # kernel_all.extend(kernel_tmp)

# sns.set(style="whitegrid", context="talk")
# plt.figure(figsize=(7, 5))
plt.figure(figsize=(8, 5))

# sns.histplot(kernel_all, bins=50, kde=False, color="steelblue", edgecolor="black")
sns.histplot(
        kernel_all, bins=50, color="#4C72B0", edgecolor="white",
        alpha=0.85, kde=True, linewidth=1.2
    )
# Add median line for context
plt.axvline(
    np.median(kernel_all), color="#E15759", linestyle="--",
    linewidth=2, label=f"Median = {np.median(kernel_all):.1f}"
)

# Add labels and styling
# plt.xlabel("Kernel Weights", fontsize=24)
# plt.ylabel("Frequency", fontsize=24)
# plt.title(dataset + ' Same Slice Kernel', fontsize=24)
# plt.xticks(fontsize=24)
# plt.yticks(fontsize=24)
# # plt.tight_layout()
# plt.show()

plt.title("Moffitt2018 Within Slice", fontsize=30, weight='bold', pad=15)
plt.xlabel("Kernel Weights", fontsize=30)
plt.ylabel("Frequency", fontsize=30)


# plt.xlabel("Number of Neighbors per Spot", fontsize=24)
# plt.ylabel("Frequency", fontsize=24)
# plt.title(dataset, fontsize=24)
plt.xticks(fontsize=30)
plt.yticks(fontsize=30)

plt.tight_layout()

# if save_path is not None:
#     fig.savefig(save_path, dpi=1200, bbox_inches='tight')

# plt.show()
save_path = os.path.join(figdir, f"{dataset}_neighbors_histogram.pdf")
plt.savefig(save_path, dpi=1200, bbox_inches="tight", transparent=True)

plt.show()
print(f"Saved: {save_path}")


```
