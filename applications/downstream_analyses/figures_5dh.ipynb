```python
import numpy as np
from sklearn.neighbors import NearestNeighbors

def _otsu_threshold(vals):
    """Simple Otsu for [0,1] floats without skimage."""
    x = np.clip((vals * 255).astype(np.uint8), 0, 255)
    hist = np.bincount(x, minlength=256).astype(float)
    prob = hist / hist.sum()
    omega = np.cumsum(prob)
    mu = np.cumsum(prob * np.arange(256))
    mu_t = mu[-1]
    denom = omega * (1.0 - omega)
    denom[denom == 0] = 1e-12
    sigma_b2 = (mu_t * omega - mu) ** 2 / denom
    t = np.argmax(sigma_b2) / 255.0
    return t

def segment_tumor_regions(
    x, y, values,
    threshold=0.1, # "otsu",            # "otsu" | ("quantile", q) | float (fixed threshold)
    k_graph=6,                   # k for kNN graph connectivity
    radius=None,                 # optional max edge length; if None auto from NN distances
    min_size=1,                 # drop components with < min_size spots
    smooth_k=0                   # k>0 to smooth values by kNN averaging before threshold
):
    """
    Parameters
    ----------
    x, y : (N,) arrays of coordinates
    values : (N,) array in [0,1] of tumor scores per spot
    threshold : see above
    k_graph : neighbors per node for connectivity
    radius : if set, keep edges <= radius (in coord units)
    min_size : minimum component size to keep
    smooth_k : if >0, replace values with kNN-averaged values (denoise)

    Returns
    -------
    labels : (N,) int array. -1 for background, 1..R for region ids
    regions : list of dicts with keys: 'id','indices','size','centroid'
    thr : numeric threshold used
    """
    x = np.asarray(x).ravel()
    y = np.asarray(y).ravel()
    v = np.asarray(values).ravel()
    assert x.shape == y.shape == v.shape, "x, y, values must be same length"
    N = x.size

    pts = np.c_[x, y]

    # Build kNN (for smoothing and graph)
    nbrs = NearestNeighbors(n_neighbors=min(k_graph+1, max(2, N)), algorithm="auto").fit(pts)
    dists, idxs = nbrs.kneighbors(pts)  # includes self at column 0

    # Optional smoothing
    if smooth_k and smooth_k > 0:
        k_use = min(smooth_k+1, idxs.shape[1])  # include self
        v = np.mean(v[idxs[:, :k_use]], axis=1)

    # Determine threshold
    if threshold == "otsu":
        thr = _otsu_threshold(np.nan_to_num(v, nan=0.0))
    elif isinstance(threshold, tuple) and threshold[0] == "quantile":
        thr = np.quantile(v, threshold[1])
    elif isinstance(threshold, (float, int)):
        thr = float(threshold)
    else:
        raise ValueError("threshold must be 'otsu', ('quantile', q), or a float")

    active = v >= thr

    # Build adjacency among active nodes using kNN and optional radius gate
    if radius is None:
        # heuristic: use median distance to the 2nd NN (exclude self) times a factor
        base = np.median(dists[:, 2]) if dists.shape[1] > 2 else np.median(dists[:, 1])
        radius = 1.8 * base

    # Adjacency list for active nodes only
    neighbors = [[] for _ in range(N)]
    k_use = min(k_graph+1, idxs.shape[1])  # include self then skip col0
    for i in range(N):
        if not active[i]: 
            continue
        for jcol in range(1, k_use):
            j = idxs[i, jcol]
            if active[j] and dists[i, jcol] <= radius:
                neighbors[i].append(j)
                neighbors[j].append(i)

    # Connected components (BFS) over active nodes
    labels = np.full(N, -1, dtype=int)
    comp_id = 0
    for i in range(N):
        if not active[i] or labels[i] != -1:
            continue
        # BFS
        queue = [i]
        labels[i] = comp_id
        head = 0
        while head < len(queue):
            u = queue[head]; head += 1
            for wj in neighbors[u]:
                if labels[wj] == -1:
                    labels[wj] = comp_id
                    queue.append(wj)
        comp_id += 1

    # Remove tiny components
    if comp_id > 0 and min_size > 1:
        sizes = np.bincount(labels[labels >= 0], minlength=comp_id)
        remove = np.where(sizes < min_size)[0]
        if len(remove) > 0:
            mask_remove = np.isin(labels, remove)
            labels[mask_remove] = -1

        # Reindex remaining components to 1..R
        kept = np.setdiff1d(np.arange(comp_id), remove)
        remap = {old: (k+1) for k, old in enumerate(kept)}  # background stays -1
        out = np.full_like(labels, -1)
        for old, new in remap.items():
            out[labels == old] = new
        labels = out
        comp_id = len(kept)
    else:
        # shift to 1..R if we didn’t filter
        labels[labels >= 0] += 1

    # Region props
    regions = []
    for rid in range(1, comp_id+1):
        idx = np.where(labels == rid)[0]
        if idx.size == 0:
            continue
        cx, cy = x[idx].mean(), y[idx].mean()
        regions.append({"id": rid, "indices": idx, "size": idx.size, "centroid": (cx, cy)})

    return labels, regions, thr
```
